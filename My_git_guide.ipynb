# What is git?
Git is a version control system. It is a collection of programs that tracks changes to files over time; typically these files are scripts or source code. It was created by Linus Torvalds in 2005 to manage development of the Linux operating system.

### What does git do?
* **History:** Every saved change to a file (called a “commit”) is recorded along with a note of who made the change, when, and a message explaining why the change was made. You can view any earlier commit of your code and even fully restore it.
* **Branching:** You can create a separate line of work (called a “branch”) to try a feature or fix without touching the main version, then merge it back when it’s ready.
* **Collaboration:** Many people can work on the same project at once, and Git helps combine their changes and flags conflicts when two people edit the same lines.

## Git terminology
* **Repository (repo):** The project folder plus its full history. The full history is stored in a hidden .git subdirectory.
* **Commit:** A snapshot of your files at a point in time.
* **Branch:** A movable pointer to a line of commits. The default is usually 'main' for main branch.
* **Remote:** A copy of the repo hosted elsewhere, such as on GitHub.

#### Git vs GitHub
One common point of confusion: Git is the tool, while GitHub is a website that hosts Git repositories and adds features like pull requests and issue tracking. You can use Git without GitHub.

## Why are we learning git in this class?
This class is not for code developers, and you do not need to learn everything git can do. For us, git is mainly a tool for **sharing code**:

* **Back up your scripts** on GitHub, so that you can easily share your workflow and scripts alongside a publication.
* **Get code from other people** with a single command.

# Undo changes
## Discard changes you have not committed
Suppose you break a file while editing it. If the file was committed before, `git restore` puts it back the way it was in your last commit. **Be careful**: changes that were never committed cannot be recovered after this.

First, let's break the file on purpose:

````bash
echo "oops, I deleted everything" > index.html
cat index.html
````

````bash
git status
````

````bash
git restore index.html
cat index.html
````

## Go back to an older version of a file
Every commit is a saved version. `git log --oneline` shows the hash of each one, and `git show` displays what a commit changed:

````bash
git show --stat HEAD
````

To understand how this works, remember that your file exists in **two places**:

* the **working copy**: the `index.html` that you see with `ls` and `cat`, and that you edit, and
* the **commits** stored inside the hidden `.git` folder. Every commit holds a complete, compressed copy of your files, but you cannot open those copies directly. You can only ask git to show you or to restore them.

`git restore --source=HASH FILE` **copies** the version of the file from that commit over your working copy. It does not delete anything from `.git`, so your newest commit is still safe there. (`HEAD` means "the latest commit", so `HEAD~1` means "one commit before the latest one", and you do not have to look up the hash.)

Let's bring back the older version of `index.html`. Remember that `cat` only shows the working copy, so it will now show the old version:

````bash
git restore --source=HEAD~1 index.html
cat index.html
````

````bash
git show HEAD:index.html
````

`git diff` compares your working copy with the latest commit. Lines starting with `-` are in the latest commit but not in your working copy any more, and lines starting with `+` are the opposite. Here, the paragraph that we added in the second commit shows up as removed:

````bash
git diff index.html
````

````bash
git status
````

Git reports `index.html` as **modified**, because your working copy no longer matches your latest commit. You now have two choices:

* **Keep the old version:** `git add` and `git commit` it. That makes a new commit that has the old content, and the history is preserved.
* **Go back to the newest version:** discard the working copy with `git restore index.html`, which copies the file from your latest commit again. That is what we will do now.

**Warning:** `git restore` overwrites your working copy. If you had made edits that you never committed, they would be lost for good, because there is no copy of them in `.git`. 

````bash
# put the latest version back
git restore index.html
cat index.html
git status
````

# Ignoring files with `.gitignore`
Not everything in your project folder belongs in git. Git is designed for **small text files like scripts**, and you should leave out:

* temporary files and folders that programs create automatically (like `.ipynb_checkpoints/` from Jupyter),
* output files (such as `*.o* and *.e* slurm files`),
* large data files (sequencing reads, genomes, BLAST databases). GitHub will refuse files over 100 MB, and your data already has a home on Kamiak, and
* **anything private**: passwords, API keys, SSH private keys, or unpublished data. Once something has been pushed to GitHub, assume it is permanently public, even if you delete it later.

A file named `.gitignore` lists patterns for files that git should pretend do not exist. Let's create one now.

````bash
cat > .gitignore << 'EOF'
# Ignore Jupyter checkpoints
.ipynb_checkpoints/
EOF
````

(The `cat > file << 'EOF'` pattern is called a *here-document*. It writes everything you type, up to a line containing only `EOF`, into the file. `EOF` is just a word that we chose to mark the end. Any word works, as long as the closing line matches the opening one exactly. The quotes around `'EOF'` tell bash to copy the text literally, without expanding andy `$variables`.)

````bash
cat .gitignore
````

Now `git status` no longer lists `.ipynb_checkpoints/`:

````bash
git status
````

.gitignore itself is a regular file that you want to share, so add and commit it:

````bash
git add .gitignore
git commit -m "Add .gitignore"
````

### A shortcut: `git add .`
`git add .` stages **every** new and modified file in the current directory (except ignored files). It is convenient, but it will stage anything you forgot about, so look at `git status` first.

# Connecting to GitHub
So far, everything we have done is on Kamiak only. Next, we will create a copy of this repository on GitHub. This gives you a backup, and a place to share your code to others who do not have access to Kamiak.

You will need a free GitHub account at [github.com](https://github.com). If you do not have one yet, create one now. (Choose a username that you would be comfortable putting on a CV.)

## Why SSH?
When you push code to GitHub, GitHub needs to know that you are really you. GitHub does not accept account passwords for git commands any more, so you have two options: a personal access token over HTTPS, or an **SSH key**. You already know how SSH keys work: GitHub will hold your **public** key, and your computer proves it has the matching **private** key. We will use SSH.

## Step 1. Check for an SSH key on Kamiak
Here is a detail that trips people up. `git push` runs **on Kamiak**, so the SSH private key has to be on Kamiak too. The key that you use to log in to Kamiak from your laptop is stored on your laptop, and it will not work here. See what keys exist on Kamiak:

````bash
ls -l ~/.ssh
````

Look for a pair of files with the same name, one ending in `.pub`, such as `id_ed25519` and `id_ed25519.pub`. The file without `.pub` is your **private key**, and you should never copy it, email it, post it, or commit it to git. The file ending in `.pub` is the **public key**, and it is the one you give to GitHub.

**If you do not have a key pair on Kamiak**, open a **terminal** (not this notebook, because the command below asks you to type a passphrase) and make one:

    ssh-keygen

Press Enter to accept the default file location, and Enter again to leave the passphrase blank.

## Step 2. Copy your public key
Print your public key. (If your file has a different name, such as `id_rsa.pub`, change it below.) It is a single line that starts with `ssh-ed25519`. Select the **whole line** and copy it.

````bash
cat ~/.ssh/id_ed25519.pub
````

## Step 3. Add the public key to your GitHub account
In your web browser:

1. Log in to GitHub, click your profile picture (top right), and choose **Settings**.
2. In the left sidebar, choose **SSH and GPG keys**.
3. Click **New SSH key**.
4. For **Title**, enter something that tells you which computer the key is on, e.g., `Kamiak`.
5. Leave **Key type** as *Authentication Key*.
6. Paste your public key into the **Key** box, and click **Add SSH key**.


## Step 4. Test the connection
We will test the connection from a **terminal** (not this notebook), because the first connection asks you a question, and if your key has a passphrase, it asks for that too. A notebook cell cannot answer either one.

Open a terminal on Kamiak and run:

    ssh -T git@github.com

`-T` means "do not try to open a shell", because GitHub does not give shell access.

**The first time you connect to a server, ssh asks whether you trust it.** You will see something like this:

```
The authenticity of host 'github.com (140.82.112.3)' can't be established.
ED25519 key fingerprint is SHA256:...
Are you sure you want to continue connecting (yes/no/[fingerprint])?
```

The fingerprint identifies the server. Before you answer, compare it with the fingerprints that GitHub publishes at [GitHub's SSH key fingerprints](https://docs.github.com/en/authentication/keeping-your-account-and-data-secure/githubs-ssh-key-fingerprints). If they match, type `yes` and press Enter. This makes sure that you are talking to the real GitHub and not an impostor. ssh remembers the answer, so it will not ask again, and it will warn you if GitHub's key ever changes.

If it worked, you will see a message like this (with your own username):

```
Hi your-github-username! You've successfully authenticated, but GitHub does not provide shell access.
```

If you see `Permission denied (publickey)`, GitHub did not recognize your key. Go back and check that you copied the **whole** public key (and not the private key), and that you pasted it into **your** account.

# Get changes from GitHub with `git pull`
Pushing sends your commits **to** GitHub. `git pull` does the reverse: it downloads commits that are on GitHub but not on your computer yet. You need it when:

* you made changes from a different computer (say, your laptop),
* you edited a file in the GitHub web interface, or
* a collaborator pushed new commits to a repository that you share.

Let's pretend to be a collaborator by changing a file **on GitHub**:

1. In your browser, open your repository on GitHub, and click on `README.md`.
2. Click the **pencil icon** (Edit this file).
3. Add a new line of text, such as `Edited on GitHub!`.
4. Click **Commit changes...**, and then **Commit changes** again in the box that opens.

Your local copy does not know about this change yet:

````bash
cat README.md
````

Now pull it down:

````bash
git pull
````

````bash
cat README.md
````

````bash
git log --oneline --decorate
````

The commit that was made on GitHub is now in your local history.

**Tip:** make `git pull` the **first** thing you do whenever you sit down to work on a shared repository, and `git push` the last. That keeps you from editing an out-of-date copy.
